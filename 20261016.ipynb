```python
1부터 100까지의 짝수의 합을 RANGE() 의 증가값을 이용하여 구하세요.
(if문자 쓰지 말것)

정수 하나를 입력받아 그 수의 구구단을 7 x 1 = 7 형식으로 출력하세요
```

```python
even_sum = 0
for i in range(1, 101):
    if i % 2 == 0:
        even_sum += i
print(f"1부터 100까지 짝수의 합: {even_sum}")
```

```python
#양의 정수를 계속 입력받아 합계를 구하다가 0을 입력하면 반복을 끝내고 합계를 출력하세요
```

```python
sum=0
while True:
  n=int(input("양의 정수를 입력하세요:"))
  if n==0:
    break
  sum+=n
print(sum)
```
