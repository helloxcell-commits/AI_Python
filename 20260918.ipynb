```python
try:
    num_str = input("숫자를 입력하세요: ")
    num = int(num_str)

if num % 2 == 0:
 print(f"{num}은(는) 짝수입니다.")
else:
 print(f"{num}은(는) 홀수입니다.")
except ValueError:
 print("유효한 숫자를 입력해주세요.")
    
```
