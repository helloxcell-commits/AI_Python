가로 12, 새로 7인 직사각형의 넓이와 둘레를 구하시오.
국어 88 \, 영어 92, 수학 75점인 학생의 과목 평균을 구하시오.

```python
가로 = 12
새로 = 7

넓이 = 12 * 7
둘레 = 2 * (12 + 7)

print(f"사각형의 넓이: {넓이}")
print(f"사각형의 둘레: {둘레}")
```

```python
국어_점수 = 88
영어_점수 = 92
수학_점수 = 75

합친_점수 = 국어_점수 + 영어_점수 + 수학_점수
과목_수 = 3
평균_점수 = 합친_점수 / 과목_수

print(f"평균_점수 : {평균_점수}점")
```
